```python
class Vehicle:
    wheels = 4

    def __init__(self, make, model):
        self.make = make
        self.model = model
    
    def description(self):
        return f"The {self.make} {self.model}"
```

```python
car = Vehicle("BMW", "i3")
```

```python
print(car.wheels)
print(car.description())
```

```python
class Budget:

    def __init__(self, budget):
        self.budget = budget

    def expense(self, amount):
        self.budget = self.budget - amount
        self.report()

    def report(self):
        print(f"Budget left:{self.budget}")
```

```python
april_budget = Budget(400)
april_budget.expense(20)
```

```python
print("Current budget variable:", april_budget.budget)
```

```python
class Budget:

    def __init__(self, budget):
        self.budget = budget

    def expense(self, amount):
        self.budget = self.budget - amount
        self.report()

    def report(self, currency = "₹"):
        print(f"Budget left:{currency}{self.budget}")
```

```python
jan_budget = Budget(23)
jan_budget.report()
jan_budget.report("$")
```
