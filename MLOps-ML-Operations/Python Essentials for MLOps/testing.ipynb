### Quick Reference Table of `unittest` Assertions

| Method | Equivalent / Description |
| --- | --- |
| `self.assertEqual(a, b)` | `a == b` |
| `self.assertNotEqual(a, b)` | `a != b` |
| `self.assertTrue(x)` | `bool(x) is True` |
| `self.assertFalse(x)` | `bool(x) is False` |
| `self.assertIs(a, b)` | `a is b` |
| `self.assertIsNot(a, b)` | `a is not b` |
| `self.assertIsNone(x)` | `x is None` |
| `self.assertIsNotNone(x)` | `x is not None` |
| `self.assertIn(a, b)` | `a in b` |
| `self.assertNotIn(a, b)` | `a not in b` |
| `self.assertIsInstance(a, b)` | `isinstance(a, b)` |
| `self.assertNotIsInstance(a, b)` | `not isinstance(a, b)` |
| `self.assertAlmostEqual(a, b)` | `round(a-b, 7) == 0` |
| `self.assertCountEqual(a, b)` | `a` and `b` have the same elements (order ignored) |

````python
import unittest

class TestAllAssertionsDemo(unittest.TestCase):

    def test_basic_equality(self):
        # assertEqual: Checks if first == second
        self.assertEqual(2 + 2, 4)
        
        # assertNotEqual: Checks if first != second
        self.assertNotEqual(2 + 2, 5)

    def test_approximate_equality(self):
        # assertAlmostEqual: Useful for floats prone to rounding issues
        self.assertAlmostEqual(0.1 + 0.2, 0.3, places=5)
        
        # assertNotAlmostEqual: Inverse of almost equal
        self.assertNotAlmostEqual(0.1 + 0.2, 0.5, places=5)
        
        # Using delta instead of places
        self.assertAlmostEqual(10.0, 10.2, delta=0.5)

    def test_container_and_sequence_equality(self):
        # assertListEqual: Specifically for lists (provides great diff output on failure)
        self.assertListEqual([1, 2, 3], [1, 2, 3])
        
        # assertTupleEqual: Specifically for tuples
        self.assertTupleEqual((1, 2), (1, 2))
        
        # assertSetEqual: Specifically for sets
        self.assertSetEqual({1, 2, 3}, {3, 1, 2})
        
        # assertDictEqual: Specifically for dictionaries
        self.assertDictEqual({'a': 1, 'b': 2}, {'b': 2, 'a': 1})
        
        # assertCountEqual: Checks if two collections have the same elements with the same counts, regardless of order
        self.assertCountEqual([1, 2, 2, 3], [3, 1, 2, 2])

    def test_identity_and_truth(self):
        a = [1, 2, 3]
        b = a
        c = [1, 2, 3]
        
        # assertIs: Checks object identity (is)
        self.assertIs(a, b)
        
        # assertIsNot: Checks that objects are not the same instance (is not)
        self.assertIsNot(a, c)
        
        # assertTrue / assertFalse: Checks boolean truthiness
        self.assertTrue(bool("Hello"))
        self.assertFalse(bool(""))

    def test_custom_failure_messages(self):
        # All assertion methods accept an optional 'msg' parameter for custom error messages
        is_admin = True
        self.assertTrue(is_admin, msg="User should have administrative privileges.")


if __name__ == '__main__':
    unittest.main(argv=[''], verbosity=2,exit=False)
````
