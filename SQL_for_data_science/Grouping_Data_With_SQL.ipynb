# Create Employees Table

````python
%%sql
DROP TABLE IF EXISTS employees;

CREATE TABLE employees (
    emp_id INT,
    name VARCHAR(50),
    department VARCHAR(50),
    salary INT,
    city VARCHAR(50)
);
````

# Insert Sample Data

````python
%%sql
INSERT INTO employees VALUES
(1,'Alice','IT',50000,'Mumbai'),
(2,'Bob','IT',60000,'Mumbai'),
(3,'Charlie','HR',45000,'Delhi'),
(4,'David','HR',55000,'Delhi'),
(5,'Eva','Sales',70000,'Bangalore'),
(6,'Frank','Sales',65000,'Mumbai');
````

# View Data

````python
%%sql
SELECT * FROM employees;
````

# COUNT Employees by Department

````python
%%sql
SELECT department,
       COUNT(*) AS employee_count
FROM employees
GROUP BY department;
````

# Average Salary by Department

````python
%%sql
SELECT department,
       AVG(salary) AS avg_salary
FROM employees
GROUP BY department;
````

# Total Salary by Department

````python
%%sql
SELECT department,
       SUM(salary) AS total_salary
FROM employees
GROUP BY department;
````

# Maximum Salary by Department

````python
%%sql
SELECT department,
       MAX(salary) AS highest_salary
FROM employees
GROUP BY department;
````

# Minimum Salary by Department

````python
%%sql
SELECT department,
       MIN(salary) AS lowest_salary
FROM employees
GROUP BY department;
````

# Group By Multiple Columns

````python
%%sql
SELECT department,
       city,
       COUNT(*) AS total_employees
FROM employees
GROUP BY department, city;
````

# HAVING Clause

````python
%%sql
SELECT department,
       COUNT(*) AS emp_count
FROM employees
GROUP BY department
HAVING COUNT(*) > 1;
````

# Difference Between WHERE and HAVING
Think of SQL processing data in this order:
- FROM
- WHERE
- GROUP BY
- HAVING
- SELECT
- ORDER BY

1. WHERE → Filters Rows Before Grouping.
WHERE works on individual rows.

````python
%%sql
SELECT *
FROM employees
WHERE salary > 50000;
````

2. HAVING → Filters Groups After Grouping.
HAVING works on the groups created by GROUP BY.


````python
%%sql
SELECT department,
       COUNT(*) AS employee_count
FROM employees
GROUP BY department
HAVING COUNT(*) > 2;
````
