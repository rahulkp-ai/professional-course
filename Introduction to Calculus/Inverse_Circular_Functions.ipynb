# Introduction to Calculus: Inverse Circular Functions

This notebook explores **Inverse Circular Functions** (Inverse Trigonometric Functions). Since standard trigonometric functions ($\sin(x)$, $\cos(x)$, $\tan(x)$) are periodic, they are **not one-to-one** and fail the horizontal line test. To define their inverses as proper functions, we must **restrict their domains**.

### Objectives:
1. Understand and visualize domain restrictions for $\sin(x)$, $\cos(x)$, and $\tan(x)$.
2. Plot and analyze the primary inverse circular functions: $\arcsin(x)$, $\arccos(x)$, and $\arctan(x)$.
3. Compute and visualize the derivatives of these inverse functions.

```python
import numpy as np
import matplotlib.pyplot as plt

# Set plotting style for clear calculus visualizations
plt.rcParams['figure.figsize'] = (12, 5)
plt.rcParams['grid.alpha'] = 0.5
plt.style.use('seaborn-v0_8-whitegrid')
```

## 1. The Inverse Sine Function: $y = \arcsin(x)$ or $y = \sin^{-1}(x)$

To invert $y = \sin(x)$, we restrict its domain to $[-\frac{\pi}{2}, \frac{\pi}{2}]$. 
On this interval, the range is $[-1, 1]$.

Consequently, for $y = \arcsin(x)$:
- **Domain:** $[-1, 1]$
- **Range:** $[-\frac{\pi}{2}, \frac{\pi}{2}]$
- **Derivative:** $\frac{d}{dx}[\arcsin(x)] = \frac{1}{\sqrt{1 - x^2}}$

```python
# 1. Generate data for restricted Sine
x_sin = np.linspace(-np.pi/2, np.pi/2, 200)
y_sin = np.sin(x_sin)

# 2. Generate data for Arcsine
x_arcsin = np.linspace(-1, 1, 200)
y_arcsin = np.arcsin(x_arcsin)

# Plotting the reflection over y = x
fig, ax = plt.subplots(1, 2, figsize=(14, 6))

# Left Subplot: Restricted Sine
ax[0].plot(x_sin, y_sin, color='blue', label=r'$y = \sin(x)$')
ax[0].set_title('Restricted Sine Function Domain')
ax[0].set_xlabel('x (radians)')
ax[0].set_ylabel('y')
ax[0].axhline(0, color='black',linewidth=0.5)
ax[0].axvline(0, color='black',linewidth=0.5)
ax[0].set_xticks([-np.pi/2, 0, np.pi/2])
ax[0].set_xticklabels([r'$-\pi/2$', '0', r'$\pi/2$'])
ax[0].legend()

# Right Subplot: Arcsine
ax[1].plot(x_arcsin, y_arcsin, color='darkblue', label=r'$y = \arcsin(x)$')
ax[1].set_title('Inverse Sine Function (Arcsine)')
ax[1].set_xlabel('x')
ax[1].set_ylabel('y (radians)')
ax[1].axhline(0, color='black',linewidth=0.5)
ax[1].axvline(0, color='black',linewidth=0.5)
ax[1].set_yticks([-np.pi/2, 0, np.pi/2])
ax[1].set_yticklabels([r'$-\pi/2$', '0', r'$\pi/2$'])
ax[1].legend()

plt.tight_layout()
plt.show()
```

## 2. The Inverse Cosine Function: $y = \arccos(x)$ or $y = \cos^{-1}(x)$

To invert $y = \cos(x)$, we restrict its domain to $[0, \pi]$. 
On this interval, the range is $[-1, 1]$.

Consequently, for $y = \arccos(x)$:
- **Domain:** $[-1, 1]$
- **Range:** $[0, \pi]$
- **Derivative:** $\frac{d}{dx}[\arccos(x)] = -\frac{1}{\sqrt{1 - x^2}}$

```python
# 1. Generate data for restricted Cosine
x_cos = np.linspace(0, np.pi, 200)
y_cos = np.cos(x_cos)

# 2. Generate data for Arccosine
x_arccos = np.linspace(-1, 1, 200)
y_arccos = np.arccos(x_arccos)

fig, ax = plt.subplots(1, 2, figsize=(14, 6))

# Left Subplot: Restricted Cosine
ax[0].plot(x_cos, y_cos, color='green', label=r'$y = \cos(x)$')
ax[0].set_title('Restricted Cosine Function Domain')
ax[0].set_xlabel('x (radians)')
ax[0].set_ylabel('y')
ax[0].axhline(0, color='black',linewidth=0.5)
ax[0].axvline(0, color='black',linewidth=0.5)
ax[0].set_xticks([0, np.pi/2, np.pi])
ax[0].set_xticklabels(['0', r'$\pi/2$', r'$\pi$'])
ax[0].legend()

# Right Subplot: Arccosine
ax[1].plot(x_arccos, y_arccos, color='darkgreen', label=r'$y = \arccos(x)$')
ax[1].set_title('Inverse Cosine Function (Arccosine)')
ax[1].set_xlabel('x')
ax[1].set_ylabel('y (radians)')
ax[1].axhline(0, color='black',linewidth=0.5)
ax[1].axvline(0, color='black',linewidth=0.5)
ax[1].set_yticks([0, np.pi/2, np.pi])
ax[1].set_yticklabels(['0', r'$\pi/2$', r'$\pi$'])
ax[1].legend()

plt.tight_layout()
plt.show()
```

## 3. The Inverse Tangent Function: $y = \arctan(x)$ or $y = \tan^{-1}(x)$

To invert $y = \tan(x)$, we restrict its domain to the open interval $(-\frac{\pi}{2}, \frac{\pi}{2})$, where the function spans all real numbers.

Consequently, for $y = \arctan(x)$:
- **Domain:** $(-\infty, \infty)$
- **Range:** $(-\frac{\pi}{2}, \frac{\pi}{2})$
- **Derivative:** $\frac{d}{dx}[\arctan(x)] = \frac{1}{1 + x^2}$

```python
# Generate data across a wider domain to showcase asymptotes
x_arctan = np.linspace(-10, 10, 500)
y_arctan = np.arctan(x_arctan)

plt.figure(figsize=(10, 5))
plt.plot(x_arctan, y_arctan, color='crimson', lw=2, label=r'$y = \arctan(x)$')

# Horizontal Asymptotes
plt.axhline(y=np.pi/2, color='gray', linestyle='--', alpha=0.7, label=r'$y = \pi/2$')
plt.axhline(y=-np.pi/2, color='gray', linestyle='--', alpha=0.7, label=r'$y = -\pi/2$')

plt.title('Inverse Tangent Function (Arctangent) & Asymptotes')
plt.xlabel('x')
plt.ylabel('y (radians)')
plt.yticks([-np.pi/2, -np.pi/4, 0, np.pi/4, np.pi/2], 
           [r'$-\pi/2$', r'$-\pi/4$', '0', r'$\pi/4$', r'$\pi/2$'])
plt.axhline(0, color='black',linewidth=0.5)
plt.axvline(0, color='black',linewidth=0.5)
plt.legend(loc='lower right')
plt.show()
```

## 4. Derivatives of Inverse Circular Functions

Let's look at the rate of change of these functions over their respective domains. Notice how the derivatives of $\arcsin(x)$ and $\arccos(x)$ approach infinity as $x \to \pm 1$ because the original trig functions have horizontal tangents ($0$ slope) at their domain boundary constraints.

```python
# Avoid evaluating exactly at -1 and 1 to prevent division by zero errors
x_deriv = np.linspace(-0.99, 0.99, 300)
x_tan_deriv = np.linspace(-4, 4, 300)

# Derivative formulas
d_arcsin = 1 / np.sqrt(1 - x_deriv**2)
d_arccos = -1 / np.sqrt(1 - x_deriv**2)
d_arctan = 1 / (1 + x_tan_deriv**2)

fig, ax = plt.subplots(1, 3, figsize=(18, 5))

# d/dx arcsin
ax[0].plot(x_deriv, d_arcsin, color='blue', label=r'$\frac{1}{\sqrt{1-x^2}}$')
ax[0].set_title(r'$\frac{d}{dx}[\arcsin(x)]$')
ax[0].set_ylim(0, 5)
ax[0].legend()

# d/dx arccos
ax[1].plot(x_deriv, d_arccos, color='green', label=r'$-\frac{1}{\sqrt{1-x^2}}$')
ax[1].set_title(r'$\frac{d}{dx}[\arccos(x)]$')
ax[1].set_ylim(-5, 0)
ax[1].legend()

# d/dx arctan
ax[2].plot(x_tan_deriv, d_arctan, color='crimson', label=r'$\frac{1}{1+x^2}$')
ax[2].set_title(r'$\frac{d}{dx}[\arctan(x)]$')
ax[2].set_ylim(0, 1.2)
ax[2].legend()

plt.tight_layout()
plt.show()
```
