```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
from IPython.display import display, Markdown
%matplotlib inline
```

## Interactive Unit Circle

```python
def plot_unit_circle(theta_deg):
    theta=np.radians(theta_deg)
    x=np.cos(theta)
    y=np.sin(theta)
    fig,ax=plt.subplots(figsize=(6,6))
    t=np.linspace(0,2*np.pi,500)
    ax.plot(np.cos(t),np.sin(t))
    ax.plot([0,x],[0,y])
    ax.scatter([x],[y])
    ax.set_aspect('equal')
    ax.grid(True)
    plt.show()
    print(f'θ={theta_deg:.1f}°')
    print(f'sin={y:.4f}')
    print(f'cos={x:.4f}')
interact(plot_unit_circle, theta_deg=FloatSlider(min=0,max=360,step=1,value=45));
```

## Sine, Cosine and Tangent Graph Explorer

```python
x=np.linspace(-2*np.pi,2*np.pi,1000)
plt.figure(figsize=(12,5))
plt.plot(x,np.sin(x),label='sin(x)')
plt.plot(x,np.cos(x),label='cos(x)')
plt.legend(); plt.grid(); plt.show()
```

## Calculus Connection: Derivative of sin(x)

```python
x=np.linspace(-2*np.pi,2*np.pi,1000)
plt.figure(figsize=(12,5))
plt.plot(x,np.sin(x),label='sin(x)')
plt.plot(x,np.cos(x),label="d/dx sin(x)=cos(x)")
plt.legend(); plt.grid(); plt.show()
```
